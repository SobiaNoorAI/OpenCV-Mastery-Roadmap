```python
#perspective transform
#points in the original image
pts1 = np.float32([[39,38], [93, 36],  [39,55], [94,50]])
#points in the output image
# height, width of output image is 300x300
pts2 = np.float32([[0,0], [300,0], [0,300], [300,300]])
    
# get the perspective transform matrix    
matrix=cv.getPerspectiveTransform(pts1, pts2) 
img_perspective = cv.warpPerspective(img, matrix, (300,300)) # apply perspective transform

#show the perspective transformed image in matplotlib
plt.imshow(cv.cvtColor(img_perspective, cv.COLOR_BGR2RGB))
plt.title('Perspective Transform')
plt.show()

# show the perspective transformed image in a window
# cv.imshow('Perspective Transform', img_perspective)
# cv.imshow('Original Image', img)
# cv.waitKey(0)
# cv.destroyAllWindows()
```

# Morphology

## Erosion

shrink the white forehead region in a binary image

in other words, it reduced the size of objects

- Eroded pixel= 255,  if at least one pixel under kernel= 255
- Eroded pixel= 0,    if any pixel under kernel < 255


```python
# erosion
marvel= cv.imread('../images/atm.png')
# convert the image to grayscale
gray_marvel=cv.cvtColor(marvel, cv.COLOR_BGR2GRAY)
# apply binary thresholding to get a binary image
binary_marvel = cv.threshold(gray_marvel, 127, 255, cv.THRESH_BINARY)
kernel = np.ones((5,5), np.uint8)
eroded_image = cv.erode(binary_marvel[1], kernel, iterations=1)

#show the eroded image in matplotlib
plt.imshow(cv.cvtColor(eroded_image, cv.COLOR_BGR2RGB))
plt.title('Erosion')
plt.show()

plt.imshow(binary_marvel[1], cmap='gray')
plt.title('Binary Image')
plt.show()
#show the eroded image in a window
# cv.imshow('Erosion', eroded_image)
# cv.imshow('Original Image', marvel)
# cv.waitKey(0)
# cv.destroyAllWindows()
```

## Dilation

Opposite of erosion

It grows the white forehead region

in other words, it increase the size of objects

- dilated pixel= 255, if at least one pixel under kernel= 255
- dilated pixel= 0, if any pixel under kernel < 255


```python
# dilation
kernel = np.ones((5,5), np.uint8)
#                           cv.dilate(src, kernel, iterations)
dilated_image = cv.dilate(binary_marvel[1], kernel, iterations=1)
#show the dilated image in matplotlib
plt.imshow(cv.cvtColor(dilated_image, cv.COLOR_BGR2RGB))
plt.title('Dilation')
plt.show()

plt.imshow(binary_marvel[1], cmap='gray')
plt.title('Binary Image')
plt.show()

# #show the dilated image in a window
# cv.imshow('Dilation', dilated_image)
# cv.imshow('Binary Image', binary_marvel[1])
# cv.waitKey(0)
# cv.destroyAllWindows()
```

## Opening

It removes small white or bright noise without affecting the shape and size of large white objects

Helps in separating connected objects in an image

Errosion--> Dialation = Opening

```python
# opening   
kernel = np.ones((5,5), np.uint8)

noise_image= cv.imread('../images/a.png')
#                               cv.morphologyEx(src, op, kernel)
opened_image = cv.morphologyEx(noise_image, cv.MORPH_OPEN, kernel)
# show the opened image in matplotlib
plt.imshow(cv.cvtColor(opened_image, cv.COLOR_BGR2RGB))
plt.title('Opening')
plt.show()

plt.imshow(noise_image)
plt.title('Noise Image')
plt.show()

#show the opened image in a window
# cv.imshow('Opening', opened_image)
# cv.imshow('Binary Image', binary_marvel[1])
# cv.waitKey(0)
# cv.destroyAllWindows()
```

 ## Closing

 Closes small holes or gaps in the foreground or white regions in a binary imag 
 
 Connects or fills small breaks in the objects

 Dilation --> Errosion = Closing

```python
# closing
kernel = np.ones((5,5), np.uint8)
noise_img=noise_image.copy()
#              cv.mophologyEx(src, op, kernel)
closed_image = cv.morphologyEx(noise_img, cv.MORPH_CLOSE, kernel)
# show the closed image in matplotlib
plt.imshow(closed_image)
plt.title('Closing')
plt.show()

plt.imshow(noise_img)
plt.title('Noisy Image')
plt.show()

# show the closed image in a window
# cv.imshow('Closing', closed_image)
# cv.imshow('Original Image', marvel)
# cv.waitKey(0)
# cv.destroyAllWindows()
```

# Contours

Edges of objects in the image

- used to detect objects, shape, and measure their size
- work on binary images because it provides sharp edges

findContours(): 

find the edges, take 3 arguments

- Source Image: image from which we want to find the contours
- Contour Retrieval Mode: This determines how contours are retrieved
 - - 1.	Retrieve External: output only the external contours
 - - 2.	Retrieve List: output all contours without any hierarchical relationship
 - - 3.	Retrieve Tree: output all contours with hierarchical relationship


Contour Approximation Method: 

This decides how much detail to keep when storing the contours
- - 1.	CHAIN_APPROX_NONE: all points are stored, no compression
- - 2.	CHAIN_APPROX_SIMPLE: store some points in straight objects like rectangle need 4 points
- - 3.	CHAIN_APPROX_TCB89_L1
- - 4.	CHAIN_APPROX_TCB89_COS


Output

The function gives us three outputs:

- Image: The image with contours found in it
- Contours: A list of contours. Each contour is made up of the (x, y) coordinates that outline a shape in the image
- Hierarchy: This gives extra information about the contours like which ones are inside others

drawContours()

Draw the contours on original image to visualize the image

```python
#contours
img_jerry=cv.imread('H:/OpenCV-Mastery-Roadmap/images/jerry.png')
gray_for_contours = cv.cvtColor(img_jerry, cv.COLOR_BGR2GRAY)
binary_for_contours = cv.threshold(gray_for_contours, 0, 255, cv.THRESH_BINARY+cv.THRESH_OTSU)

#cv.findContours(image, mode, method)
contours, hierarchy = cv.findContours(binary_for_contours[1], cv.RETR_TREE, cv.CHAIN_APPROX_SIMPLE)
print(f'Number of contours found: {len(contours)}')

# draw contours on the original image
# cv.drawContours(image, contours, contourIdx, color, thickness)
jerry_with_contours=cv.drawContours(img_jerry, contours, -1, (0,255,0), 3)

# show the image with contours in matplotlib
# show only contour edges on a black background
edges_only = np.zeros_like(img_jerry) # create a black image with the same shape as the original
edges_only = cv.drawContours(edges_only, contours, -1, (0,255,0), 3)
plt.imshow(cv.cvtColor(edges_only, cv.COLOR_BGR2RGB))
plt.title('Contours edges only')
plt.axis('off')
plt.show()

plt.imshow(cv.cvtColor(jerry_with_contours, cv.COLOR_BGR2RGB))
plt.title('Contours image')
plt.show()



# cv.imshow('Contours', jerry_with_contours)
# cv.imshow('Gray', gray_for_contours)
# cv.imshow('Binary', binary_for_contours[1])
# cv.waitKey(0)
# cv.destroyAllWindows()
```
