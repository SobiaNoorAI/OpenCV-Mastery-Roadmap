# Color Spaces

You have notice that when we display image of jerry using matplotlib, its color changes to blue. It is because maplotlib library works on RGB color space & OpenCV works on BGR. So, if we want to disply orginal color of Jerry, we need to convert the color space using cv.cvtColor()

```python
img=jerry.copy()
rgb_img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.imshow(rgb_img)
plt.title('BGR in matplotlib')
```

### Display All Color Spaces in OpenCV

```python
# Convert to various color spaces
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)
lab = cv.cvtColor(img, cv.COLOR_BGR2LAB)
ycrcb = cv.cvtColor(img, cv.COLOR_BGR2YCrCb)

# 2. Define dimensions for a 2x3 grid to fit 1280x720
# Width: 1280 / 3 ≈ 426 | Height: 720 / 2 = 360
target_w, target_h = 426, 360

# Resize the source image first
img_resized = cv.resize(img, (target_w, target_h))
# Display each in a unique window
cv.imshow('Original BGR', img)
cv.imshow('Grayscale', gray)
cv.imshow('HSV Space', hsv)
cv.imshow('LAB Space', lab)
cv.imshow('YCrCb Space', ycrcb)

# Standard OpenCV cleanup
cv.waitKey(0)
cv.destroyAllWindows()
```

Display All color spaces in a Grid View

```python
img_opencv= jerry.copy()
# 1. Dimensions for 2x3 grid to fit 1280x720 (Width: 426, Height: 360)
target_w, target_h = 426, 360

def prepare_image(image, label, is_gray=False):
    # Resize first
    res = cv.resize(image, (target_w, target_h))
    
    # If grayscale, convert to 3-channel BGR for stacking
    if is_gray:
        res = cv.cvtColor(res, cv.COLOR_GRAY2BGR)
    
    # Add Label (Bright Green Text)
    cv.putText(res, label, (20, 45), cv.FONT_HERSHEY_SIMPLEX, 
                1.1, (0, 255, 0), 2, cv.LINE_AA)
    return res

# 2. Create all 6 Color Spaces
bgr   = prepare_image(img_opencv, "BGR (OpenCV Default)")
rgb   = prepare_image(cv.cvtColor(img_opencv, cv.COLOR_BGR2RGB), "RGB (Swapped)")
gray  = prepare_image(cv.cvtColor(img_opencv, cv.COLOR_BGR2GRAY), "Grayscale", is_gray=True)
hsv   = prepare_image(cv.cvtColor(img_opencv, cv.COLOR_BGR2HSV), "HSV")
lab   = prepare_image(cv.cvtColor(img_opencv, cv.COLOR_BGR2LAB), "LAB")
ycrcb = prepare_image(cv.cvtColor(img_opencv, cv.COLOR_BGR2YCrCb), "YCrCb")

# 3. Stack and Display
# Top Row: BGR, RGB, Grayscale
top_row = np.hstack((bgr, rgb, gray))
# Bottom Row: HSV, LAB, YCrCb
bottom_row = np.hstack((hsv, lab, ycrcb))

final_grid = np.vstack((top_row, bottom_row))

# 4. Show Final Result
cv.imshow('6 Color Spaces Grid - 1280x720', final_grid)
cv.waitKey(0)
cv.destroyAllWindows()
```

### Display All Color Spaces in Matplotlib

```python
# 1. Define the spaces and their names
# We pre-convert them to ensure they display correctly in the grid
data = [
    ("1. Original BGR", jerry_img, None),
    ("2. Corrected RGB", cv.cvtColor(jerry_img, cv.COLOR_BGR2RGB), None),
    ("3. Grayscale", cv.cvtColor(jerry_img, cv.COLOR_BGR2GRAY), 'gray'),
    ("4. HSV Space", cv.cvtColor(jerry_img, cv.COLOR_BGR2HSV), None),
    ("5. LAB Space", cv.cvtColor(jerry_img, cv.COLOR_BGR2LAB), None),
    ("6. YCrCb Space", cv.cvtColor(jerry_img, cv.COLOR_BGR2YCrCb), None)
]

# 2. Create a 2x3 grid
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('OpenCV Color Space Comparison in Matplotlib', fontsize=16)

# 3. Flatten the axes array to iterate easily
axes = axes.flatten()

for i, (title, img, cmap) in enumerate(data):
    if cmap:
        axes[i].imshow(img, cmap=cmap)
    else:
        axes[i].imshow(img)
    
    axes[i].set_title(title)
    axes[i].axis('off') # Hide tick marks

plt.tight_layout() # Adjust spacing
plt.show()
```
